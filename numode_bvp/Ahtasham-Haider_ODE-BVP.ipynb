## **Mathematical Formulation of the Blasius Boundary Layer Equation**

For a steady, incompressible flow of a uniform stream $U_\infty$ over a flat plate, applying the similarity transformation:
$$\eta = y\sqrt{\frac{U_\infty}{\nu x}}, \quad \frac{u}{U_\infty} = f'(\eta)$$

reduces the governing boundary layer equations to the third-order nonlinear ordinary differential equation known as the **Blasius equation**:
$$f''' + \frac{1}{2}ff'' = 0, \quad 0 < \eta < \infty \quad \text{--- (1)}$$

This equation is subject to the following boundary conditions:
* **At the wall ($\eta = 0$):** $f(0) = 0$ and $f'(0) = 0$ *(no-slip)*
* **In the free stream ($\eta \to \infty$):** $f'(\eta) \to 1$ *(free-stream)*


## **Reduction to a First-Order System**

To solve this third-order boundary value problem numerically, it is converted into a system of three coupled first-order ordinary differential equations. We define the state variables as follows:
$$Y_1 = f$$
$$Y_2 = Y_1' = f'$$
$$Y_3 = Y_2' = f''$$

Substituting these definitions into Equation (1) yields:
$$Y_3' + \frac{1}{2}Y_1Y_3 = 0$$

Rearranging and combining with the state definitions gives the complete system of first-order equations:
$$Y_1' = Y_2 \quad \text{--- (2)}$$
$$Y_2' = Y_3 \quad \text{--- (3)}$$
$$Y_3' = -\frac{1}{2}Y_1Y_3 \quad \text{--- (4)}$$

---

#### **Matrix Representation**

This system can be expressed in vector-matrix form as:
$$\mathbf{Y}' = \mathbf{G}(\mathbf{Y})$$

where
$$\mathbf{Y}' = \begin{bmatrix} Y_1' \\ Y_2' \\ Y_3' \end{bmatrix}, \quad \mathbf{G}(\mathbf{Y}) = \begin{bmatrix} Y_2 \\ Y_3 \\ -\frac{1}{2}Y_1Y_3 \end{bmatrix}$$

## **Numerical Method**

#### **Standard RK3 Formula**

Given a step size h (Δη) and a current point $(t_n, y_n)$, the next value $y_{n+1}$ is computed as:

$$\mathbf{y_{n+1}} = \mathbf{y_n} + \mathbf{\frac{h}{6}(k_1 + 4k_2 + k_3)}$$

Where the three slope increments (k-values) are defined as:

*   $k1 = f(t_n, y_n)$
*   $k2 = f(t_n + \frac{h}{2}, y_n + \frac{h}{2}k_1)$
*   $k3 = f(t_n + h, y_n - hk_1 + 2hk_2)$


#### **Standard RK4 Formula**
Given a step size $h$ ($\Delta\eta$) and a current point $(t_n, y_n)$, the next value $y_{n+1}$ is computed as:
$$\mathbf{y_{n+1}} = \mathbf{y_n} + \mathbf{\frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4)}$$

Where the four slope increments ($k$-values) are defined as:
* $k_1 = f(t_n, y_n)$
* $k_2 = f(t_n + \frac{h}{2}, y_n + \frac{h}{2}k_1)$
* $k_3 = f(t_n + \frac{h}{2}, y_n + \frac{h}{2}k_2)$
* $k_4 = f(t_n + h, y_n + hk_3)$


```python
# Runge-Kutta 3 Function
def RK_3(Y, η_span, func, Δη):

    (ηi, ηf) = η_span       # Starting and End Point

    N = int(round((ηf - ηi)/Δη))
    η_matrix = np.linspace(ηi, ηf, N+1)               # It will hold time steps
    Y_matrix = np.zeros((len(η_matrix), len(Y)))        # It will hold f, f' f'' at different time steps
    η_matrix[0] = ηi        # Initializing time steps
    Y_matrix[0] = Y         # Intial Values of f, f', f''

    for i in range(len(η_matrix) - 1):
        k1 = Δη*func(η_matrix[i], Y_matrix[i])
        k2 = Δη*func(η_matrix[i] + Δη/2, Y_matrix[i] + k1/2)
        k3 = Δη*func(η_matrix[i] + Δη, Y_matrix[i] - k1 + 2*k2)
        Y_matrix[i+1] = Y_matrix[i] + (k1 + 4*k2 + k3)/6
        
    return η_matrix, Y_matrix


# Runge-Kutta 4 Function
def RK_4(Y, η_span, func, Δη):

    (ηi, ηf) = η_span       # Starting and End Point

    N = int(round((ηf - ηi)/Δη))
    η_matrix = np.linspace(ηi, ηf, N+1)               # It will hold time steps
    Y_matrix = np.zeros((len(η_matrix), len(Y)))        # It will hold f, f', f'' at different time steps
    η_matrix[0] = ηi        # Initializing time steps
    Y_matrix[0] = Y         # Initial Values of f, f', f''

    for i in range(len(η_matrix) - 1):
        k1 = Δη*func(η_matrix[i], Y_matrix[i])
        k2 = Δη*func(η_matrix[i] + Δη/2, Y_matrix[i] + k1/2)
        k3 = Δη*func(η_matrix[i] + Δη/2, Y_matrix[i] + k2/2)
        k4 = Δη*func(η_matrix[i] + Δη, Y_matrix[i] + k3)
        Y_matrix[i+1] = Y_matrix[i] + (k1 + 2*k2 + 2*k3 + k4)/6
        
    return η_matrix, Y_matrix
```

```python
# Derivative Vector: DY_DX = [Y_1', Y_2', Y_3']
def DY_DX(η, Y):

    return np.array([Y[1], Y[2], -0.5*Y[0]*Y[2]])       # return G(Y) = [Y_2, Y_3, -1/2*Y_1*Y_3]
```

### **Residual Function**

$$\mathbf{F(s, η∞)} = \mathbf{f'(s, η∞)} - 1$$

```python
# Residual Function →  F(s;η∞) = f′(η∞;s)−1.
def F(s, h, method):

    if method=="RK-3":
        Y = np.array([f0, df0, s])
        _, f = RK_3(Y, (ηi, ηf), DY_DX, h)     # f′(η∞;s) = f[-1, 1]
    else:
        Y = np.array([f0, df0, s])
        _, f = RK_4(Y, (ηi, ηf), DY_DX, h)
        
    return f[-1, 1] - 1
```

### **Shooting method**


*   Turn the BVP into a sequence of IVPs.
*   Guess the missing slope $s = f′′(0)$.
*   Solve the IVP
      $$y′′ = f (x,y,y′), y(a) = α, y′(a) = s$$
        with an IVP solver Call the solution y(x;s).

*   Compare the end value with the target: define the residual
        $$F(s) = y(b;s) −β$$
*   Adjust s until F(s) = 0 ⇒ This is a root-finding problem

```python
# Shooting Method
def shooting_method(R, S, h, method, tol=1e-6, max_iter=1000, show_param=False):
    
    xL, xR = S               # Initial Guesses
    R = R                    # Function
    tol = tol                # Tolerance
    max_iter = max_iter      # Maximum Iteration
    m = method               # method = ""RK-3"  or "RK-4" 

    i = 0                    # Iteration
    err = abs(xR - xL)       # Initial Error


    # Bisection Method:
    if R(xL, h, m)*R(xR, h, m) > 0:
        raise ValueError("f(xL) and f(xR) must have opposite signs")
    
    while tol < err and i < max_iter:
        if R(xL, h, m) == 0:                  # if True xL is the root
            return xL
        elif R(xR, h, m) == 0:                # if True xR is the root
            return xR
        else:
            xM = (xL + xR)/2
            if R(xM, h, m) == 0:              # if True xM is the root
                return xM
            elif R(xL, h, m)*R(xM, h, m) < 0:       # Means f(xM) and f(xR) has same sign
                xR = xM
            else:
                xL = xM
        err = abs(xR - xL)
        i+=1

    if show_param:
        print(f"Error = {err}")
        print(f"Iteration {i} completed !, \n")
        
    return (xL + xR)/2
```

```python
# Initiating Shooting Method
S = (0.1, 0.5)      # Initial guess for shooting method
s = shooting_method(F, S, Δη, method="RK-3", tol=1e-12, show_param=True)
print(f"Converged Value:→ f′′(0) = {s}")
```

## **Parametric Variation**

### **RK-3**

```python
Δη_matrix = np.array([0.4, 0.2, 0.1, 0.05, 0.025])
acc_matrix_rk3 = np.zeros_like(Δη_matrix)
s_exact = 0.332057336215196

# Initiating Shooting Method
S = (0.1, 0.5)      # Initial guess for shooting method

# Plot of f, f′ and f′′ against η
fig, ax = plt.subplots(len(Δη_matrix), 3, figsize=(12, 15), constrained_layout=True)

# Column titles 
col_titles = ["η vs f(η)", "η vs f'(η)", "η vs f''(η)"]
x_labels = ["f(η)", "f'(η) = u/V∞", "f''(η)"]

for i, Δη in enumerate(Δη_matrix):
    s = shooting_method(F, S, Δη, method="RK-3", tol=1e-12)
    acc_matrix_rk3[i] = abs(s - s_exact)

    Y = np.array([0, 0, s])
    η_matrix, Y_matrix = RK_3(Y, (ηi, ηf), DY_DX, Δη)

    for j in range(3):
        ax[i, j].plot(Y_matrix[:, j], η_matrix, linewidth=1.8)
        ax[i, j].grid(True, linestyle='--', alpha=0.4)
        
        # Set top titles only for the first row
        if i == 0:
            ax[i, j].set_title(col_titles[j], fontsize=13, fontweight='bold', pad=10)
            
        # Set x-labels only for the bottom row (or keep all if preferred)
        if i == len(Δη_matrix) - 1:
            ax[i, j].set_xlabel(x_labels[j], fontsize=11, fontweight='bold')

    # Include step size in the left-most y-label so each row is distinct
    ax[i, 0].set_ylabel(f"Δη = {Δη}\n\nη", fontsize=11, fontweight='bold')

fig.suptitle("Blasius Boundary Layer Profiles across Step Sizes (RK-3)", fontsize=16, fontweight='bold')
plt.show()
```

### **RK-4**

```python
Δη_matrix = np.array([0.4, 0.2, 0.1, 0.05, 0.025])
acc_matrix_rk4 = np.zeros_like(Δη_matrix)
s_exact = 0.332057336215196

# Initiating Shooting Method
S = (0.1, 0.5)      # Initial guess for shooting method

# Plot of f, f′ and f′′ against η
fig, ax = plt.subplots(len(Δη_matrix), 3, figsize=(12, 15), constrained_layout=True)

# Column titles 
col_titles = ["η vs f(η)", "η vs f'(η)", "η vs f''(η)"]
x_labels = ["f(η)", "f'(η) = u/V∞", "f''(η)"]

for i, Δη in enumerate(Δη_matrix):
    s = shooting_method(F, S, Δη, method="RK-4", tol=1e-12)
    acc_matrix_rk4[i] = abs(s - s_exact)

    Y = np.array([0, 0, s])
    η_matrix, Y_matrix = RK_4(Y, (ηi, ηf), DY_DX, Δη)

    for j in range(3):
        ax[i, j].plot(Y_matrix[:, j], η_matrix, linewidth=1.8)
        ax[i, j].grid(True, linestyle='--', alpha=0.4)
        
        # Set top titles only for the first row
        if i == 0:
            ax[i, j].set_title(col_titles[j], fontsize=13, fontweight='bold', pad=10)
            
        # Set x-labels only for the bottom row (or keep all if preferred)
        if i == len(Δη_matrix) - 1:
            ax[i, j].set_xlabel(x_labels[j], fontsize=11, fontweight='bold')

    # Include step size in the left-most y-label so each row is distinct
    ax[i, 0].set_ylabel(f"Δη = {Δη}\n\nη", fontsize=11, fontweight='bold')

fig.suptitle("Blasius Boundary Layer Profiles across Step Sizes (RK-4)", fontsize=16, fontweight='bold')
plt.show()
```

## **Convergence of RK-3 and RK-4**

```python
plt.figure(figsize=(6, 4))
plt.loglog(Δη_matrix, acc_matrix_rk3, 'o--', label='RK3 Error')
plt.loglog(Δη_matrix, acc_matrix_rk4, 's--', label='RK4 Error')

plt.xlabel(r'Step size $\Delta\eta$', fontweight='bold')
plt.ylabel('Absolute error in shooting parameter', fontweight='bold')
plt.title('RK3 and RK4 Convergence Study', fontweight='bold')
plt.grid(True, which='both')
plt.gca().invert_xaxis()
plt.legend()
plt.show()
```

```python
# Calculate observed order of accuracy
p_rk3 = np.log(acc_matrix_rk3[:-1] / acc_matrix_rk3[1:]) / \
        np.log(Δη_matrix[:-1] / Δη_matrix[1:])

p_rk4 = np.log(acc_matrix_rk4[:-1] / acc_matrix_rk4[1:]) / \
        np.log(Δη_matrix[:-1] / Δη_matrix[1:])

print("Observed order of accuracy for RK3:")
print(p_rk3)

print("\nObserved order of accuracy for RK4:")
print(p_rk4)
```

## **Results and Discussion**

* **Shooting Method Convergence:** Using an iterative root-finding approach, the shooting method successfully converged to the unknown initial boundary condition $f''(0) \approx 0.33206$ within **39 iterations**, closely matching the classical Blasius benchmark value ($0.332057$).
* **Step-Size Sensitivity Analysis:** Equation (1) was numerically integrated across progressively refined spatial step sizes ($\Delta\eta \in \{0.4, 0.2, 0.1, 0.05, 0.025\}$) using both the **3rd-order (RK3)** and **4th-order (RK4)** Runge–Kutta schemes. Both solvers accurately captured the boundary layer velocity profiles ($f(\eta)$, $f'(\eta)$, and $f''(\eta)$) as $\Delta\eta$ decreased.
* **Order of Accuracy Verification:** Log–log error analysis confirms that the absolute error in the shooting parameter decreases monotonically as the grid is refined. As theoretically expected, **RK3** exhibits a third-order convergence rate ($\mathcal{O}(\Delta\eta^3)$), whereas **RK4** demonstrates fourth-order convergence ($\mathcal{O}(\Delta\eta^4)$), achieving significantly lower numerical error for identical step sizes.


## **Conclusion**

* **Method Effectiveness:** Coupling the iterative shooting method with higher-order Runge–Kutta schemes provides an accurate and reliable numerical framework for solving the nonlinear Blasius boundary layer equation.
* **Solver Comparison:** While both RK3 and RK4 converge smoothly toward the exact wall shear parameter ($f''(0) \approx 0.332057$), **RK4 achieves superior accuracy** at coarser step sizes due to its fourth-order truncation error ($\mathcal{O}(\Delta\eta^4)$), making it more computationally efficient for high-precision boundary layer simulations.
