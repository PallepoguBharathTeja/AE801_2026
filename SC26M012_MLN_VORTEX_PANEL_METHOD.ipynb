## Abstract

The vortex panel method is a numerical approach used to
analyse the aerodynamic characteristics of two-dimensional
airfoils based on potential flow theory.

In this work, constant-strength and linear-strength vortex
panel methods are implemented using Python. The airfoil
surface is represented by a finite number of panels, and
the vortex strengths are determined by satisfying the
flow tangency condition at the control points.

The numerical formulation is first examined using a
circular cylinder. Subsequently, NACA 0012 and NACA 2412
airfoils are investigated to determine their pressure
coefficient and lift coefficient distributions.

A grid convergence study is also performed to examine
the influence of discretization on the numerical results.

The study provides a comparison of the two vortex panel
formulations and their application to preliminary
aerodynamic analysis.

## 1. Introduction

Aerodynamic analysis plays an important role in the
design and performance evaluation of aircraft wings
and other aerodynamic surfaces.

The vortex panel method is a computational technique
based on potential flow theory. It represents the
airfoil surface using a collection of discrete panels
over which vortex strength is distributed.

The unknown vortex strengths are calculated by applying
the no-penetration boundary condition at selected
control points on the airfoil surface.

Two formulations are considered in this study:

1. Constant-strength vortex panel method.
2. Linear-strength vortex panel method.

In the constant formulation, vortex strength remains
uniform over each panel, whereas in the linear
formulation, it varies between the panel endpoints.

These methods are applied to symmetric and cambered
airfoils to investigate their aerodynamic behaviour.

## 3. Theoretical Background

The vortex panel method is developed using the
assumptions of potential flow theory.

The main assumptions are:

- The flow is incompressible.
- The flow is inviscid.
- The flow is irrotational outside the vortex sheet.
- The flow is two-dimensional.
- The airfoil surface is impermeable.

The airfoil is divided into a number of straight panels.
Each panel contributes to the velocity field through
its associated vortex distribution.

The total velocity at a control point is obtained by
combining the free-stream velocity and the induced
velocities from all panels.

The unknown vortex strengths are determined by
satisfying the boundary conditions.

## 4. Governing Equations

### 4.1 No-penetration boundary condition

The velocity normal to the airfoil surface must be zero.

$$
V_n = 0
$$

The total velocity is the combination of free-stream
velocity and induced velocity.

### 4.2 Pressure coefficient

Using Bernoulli's equation, the pressure coefficient
is calculated as:

$$
C_p = 1 - \left(\frac{V_t}{V_\infty}\right)^2
$$

where:

- $V_t$ is the local tangential velocity.
- $V_\infty$ is the free-stream velocity.

### 4.3 Lift coefficient

The lift coefficient is obtained from the circulation:

$$
C_l = \frac{2\Gamma}{V_\infty c}
$$

where:

- $\Gamma$ is the total circulation.
- $c$ is the airfoil chord length.

For constant vortex strength:

$$
\Gamma = \sum_{j=1}^{N}\gamma_j S_j
$$

For linear vortex strength:

$$
\Gamma =
\sum_{j=1}^{N}
\frac{\gamma_j+\gamma_{j+1}}{2}S_j
$$

## 5. Numerical Methodology

The numerical analysis is carried out through the
following steps:

1. Generate the required airfoil or cylinder geometry.
2. Discretize the geometry into straight-line panels.
3. Calculate panel lengths, orientations and control points.
4. Calculate the normal and tangential influence coefficients.
5. Assemble the governing linear system.
6. Apply the Kutta condition at the trailing edge.
7. Solve for the unknown vortex strengths.
8. Calculate tangential velocity and pressure coefficient.
9. Determine the lift coefficient from total circulation.
10. Compare the numerical results for both formulations.

The implementation uses NumPy for numerical calculations
and Matplotlib for graphical representation.

## 6. Airfoil Geometry Generation

The airfoil coordinates are generated using the
NACA four-digit airfoil representation.

Two airfoils are considered:

### NACA 0012

- Symmetric airfoil.
- Maximum thickness is 12% of chord.
- Zero camber.

### NACA 2412

- Maximum camber is 2% of chord.
- Maximum camber position is at 40% of chord.
- Maximum thickness is 12% of chord.

Cosine spacing is used to distribute the coordinate
points along the chord. This provides better
resolution near the leading and trailing edges.

The upper and lower surface coordinates are then
combined to form the complete airfoil geometry.

## 7. Circular Cylinder Geometry

A circular cylinder is considered as the initial
verification geometry.

The cylinder coordinates are generated using
the parametric equations:

$$
x = R\cos\theta
$$

$$
y = -R\sin\theta
$$

where $R$ is the cylinder radius and $\theta$
is the angular coordinate.

The generated coordinates are divided into
straight-line panels for numerical analysis.

## 8. Panel Discretization

The generated geometry is divided into a finite
number of straight-line panels.

For every panel, the following properties are calculated:

- Panel length.
- Midpoint coordinates.
- Panel orientation.
- Unit normal direction.

The panel midpoint is selected as the control point.

The boundary condition is imposed at each control point
to determine the unknown vortex strengths.

The panel geometry is used by both the constant-strength
and linear-strength formulations.

## 9. Constant-Strength Vortex Panel Method

In the constant-strength formulation, the vortex
strength is assumed to remain constant over each panel.

The induced normal velocity contributions are
assembled into an influence coefficient matrix.

The resulting system can be represented as:

$$
[K]\{\gamma\}=\{B\}
$$

where:

- $[K]$ is the normal influence coefficient matrix.
- $\{\gamma\}$ is the unknown vortex strength vector.
- $\{B\}$ represents the free-stream contribution.

The Kutta condition is applied at the trailing edge:

$$
\gamma_1+\gamma_N=0
$$

The resulting system is solved to determine the
vortex strengths.

The tangential velocity and pressure coefficient
are subsequently evaluated.

## 10. Linear-Strength Vortex Panel Method

In the linear-strength formulation, the vortex strength
varies linearly between the endpoints of each panel.

The vortex strengths are defined at the panel nodes.

The unknown nodal strengths are represented by:

$$
\gamma_1,\gamma_2,\ldots,\gamma_{N+1}
$$

The normal and tangential influence coefficients
are calculated using the linear variation of
vortex strength.

The resulting system is expressed as:

$$
[A_n]\{\gamma\}=\{B\}
$$

The Kutta condition is imposed as:

$$
\gamma_1+\gamma_{N+1}=0
$$

After solving the system, the tangential velocity,
pressure coefficient and lift coefficient
are determined.

## 11. Verification Using a Circular Cylinder

The numerical formulation is initially examined
using a circular cylinder.

For potential flow around a circular cylinder
without circulation, the analytical pressure
coefficient is:

$$
C_p = 1-4\sin^2\theta
$$

where $\theta$ is the angular position measured
from the front stagnation point.

The theoretical lift coefficient is:

$$
C_l=0
$$

The numerical pressure coefficient is compared
with the analytical distribution.

This comparison provides a preliminary verification
of the numerical implementation.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class Airfoil:
    def __init__(self):
        self.name = ""
        self.x = np.array([])
        self.y = np.array([])

    def symmetric(self, t, num_points):

        beta = np.linspace(0, np.pi, num_points)

        x = 0.5 * (1 - np.cos(beta))

        y = 5*t*(0.2969*x**0.5 - 0.1260*x
                 - 0.3516*x**2 + 0.2843*x**3
                 - 0.1036*x**4)

        y_u = y
        y_l = -y

        self.x = np.concatenate((np.flip(x), x[1:]))
        self.y = np.concatenate((np.flip(y_l), y_u[1:]))

        return self.x, self.y

    def cambered(self, t, num_points):

        m = 0.02
        p = 0.4

        beta = np.linspace(0, np.pi, num_points)

        x = 0.5 * (1 - np.cos(beta))

        y = 5*t*(0.2969*x**0.5 - 0.1260*x
                 - 0.3516*x**2 + 0.2843*x**3
                 - 0.1036*x**4)

        y_c = np.where(
            x <= p,
            (m/p**2)*(2*p*x - x**2),
            (m/(1-p)**2)*(1 - 2*p + 2*p*x - x**2)
        )

        dy_dx = np.where(
            x <= p,
            (2*m/p**2)*(p-x),
            (2*m/(1-p)**2)*(p-x)
        )

        theta = np.arctan(dy_dx)

        x_u = x - y*np.sin(theta)
        x_l = x + y*np.sin(theta)

        y_u = y_c + y*np.cos(theta)
        y_l = y_c - y*np.cos(theta)

        self.x = np.concatenate((np.flip(x_l), x_u[1:]))
        self.y = np.concatenate((np.flip(y_l), y_u[1:]))

        return self.x, self.y
```

```python
class Cylinder:
    def __init__(self, radius=1.0):

        self.name = "Circular Cylinder"
        self.R = radius

        self.x = np.array([])
        self.y = np.array([])

    def generate(self, num_points=100):

        beta = np.linspace(0, 2*np.pi, num_points)

        self.x = self.R*np.cos(beta)
        self.y = -self.R*np.sin(beta)

        return self.x, self.y
```

```python
class Panel:
    def __init__(self, x_nodes, y_nodes):

        self.x = x_nodes
        self.y = y_nodes

        self.N = len(x_nodes) - 1

        self.compute_panels()

    def compute_panels(self):

        dx = self.x[1:] - self.x[:-1]
        dy = self.y[1:] - self.y[:-1]

        self.length = np.sqrt(dx**2 + dy**2)

        self.x_mid = (self.x[:-1] + self.x[1:])/2
        self.y_mid = (self.y[:-1] + self.y[1:])/2

        self.theta = np.arctan2(dy, dx)

        self.nx = -dy/self.length
        self.ny = dx/self.length
```

```python
class Vortex_Panel_Method:

    def __init__(self, Panel, V_fs=10.0, alpha=0.0):

        self.Panel = Panel
        self.V_fs = V_fs
        self.alpha = np.radians(alpha)

    # -----------------------------------------
    # CONSTANT STRENGTH VORTEX PANEL METHOD
    # -----------------------------------------

    def constant_vortex_strength(self):

        N = self.Panel.N

        K = np.zeros((N, N))
        L = np.zeros((N, N))
        B = np.zeros(N)

        for i in range(N):

            B[i] = self.V_fs*np.sin(
                self.Panel.theta[i] - self.alpha
            )

            for j in range(N):

                if i == j:

                    K[i, j] = 0
                    L[i, j] = 0.5

                else:

                    a = -(self.Panel.x_mid[i] - self.Panel.x[j])*np.cos(
                        self.Panel.theta[j]
                    ) - (self.Panel.y_mid[i] - self.Panel.y[j])*np.sin(
                        self.Panel.theta[j]
                    )

                    b = (self.Panel.x_mid[i] - self.Panel.x[j])**2 + (
                        self.Panel.y_mid[i] - self.Panel.y[j]
                    )**2

                    c = np.sin(
                        self.Panel.theta[i] - self.Panel.theta[j]
                    )

                    d = np.cos(
                        self.Panel.theta[i] - self.Panel.theta[j]
                    )

                    e = (self.Panel.x_mid[i] - self.Panel.x[j])*np.sin(
                        self.Panel.theta[j]
                    ) - (self.Panel.y_mid[i] - self.Panel.y[j])*np.cos(
                        self.Panel.theta[j]
                    )

                    f = np.log(
                        1 + (
                            self.Panel.length[j]**2
                            + 2*a*self.Panel.length[j]
                        )/b
                    )

                    g = np.arctan2(
                        e*self.Panel.length[j],
                        b + a*self.Panel.length[j]
                    )

                    K[i, j] = (0.5*d*f + c*g)/(2*np.pi)

                    L[i, j] = (0.5*c*f - d*g)/(2*np.pi)

        # Kutta condition

        K[-1, :] = 0
        K[-1, 0] = 1
        K[-1, -1] = 1

        B[-1] = 0

        # Solve for vortex strength

        self.gamma = np.linalg.solve(K, B)

        self.K = K
        self.L = L

        # Tangential velocity

        self.Vt = (
            self.V_fs*np.cos(self.Panel.theta - self.alpha)
            + self.L @ self.gamma
        )

        # Pressure coefficient

        self.Cp = 1 - (self.Vt/self.V_fs)**2

        # Lift coefficient

        c = self.Panel.x.max() - self.Panel.x.min()

        Gamma = np.sum(self.gamma*self.Panel.length)

        self.Cl = 2*Gamma/(self.V_fs*c)

        return self.Cp, self.Cl

    # -----------------------------------------
    # LINEAR STRENGTH VORTEX PANEL METHOD
    # -----------------------------------------

    def linear_vortex_strength(self):

        N = self.Panel.N

        Cn1 = np.zeros((N, N))
        Cn2 = np.zeros((N, N))

        Ct1 = np.zeros((N, N))
        Ct2 = np.zeros((N, N))

        RHS = np.zeros(N)

        for i in range(N):

            RHS[i] = self.V_fs*np.sin(
                self.Panel.theta[i] - self.alpha
            )

            for j in range(N):

                if i == j:

                    Cn1[i, j] = -1/(2*np.pi)
                    Cn2[i, j] = 1/(2*np.pi)

                    Ct1[i, j] = (np.pi/2)/(2*np.pi)
                    Ct2[i, j] = (np.pi/2)/(2*np.pi)

                else:

                    a = -(self.Panel.x_mid[i] - self.Panel.x[j])*np.cos(
                        self.Panel.theta[j]
                    ) - (self.Panel.y_mid[i] - self.Panel.y[j])*np.sin(
                        self.Panel.theta[j]
                    )

                    b = (self.Panel.x_mid[i] - self.Panel.x[j])**2 + (
                        self.Panel.y_mid[i] - self.Panel.y[j]
                    )**2

                    c = np.sin(
                        self.Panel.theta[i] - self.Panel.theta[j]
                    )

                    d = np.cos(
                        self.Panel.theta[i] - self.Panel.theta[j]
                    )

                    e = (self.Panel.x_mid[i] - self.Panel.x[j])*np.sin(
                        self.Panel.theta[j]
                    ) - (self.Panel.y_mid[i] - self.Panel.y[j])*np.cos(
                        self.Panel.theta[j]
                    )

                    f = np.log(
                        1 + (
                            self.Panel.length[j]**2
                            + 2*a*self.Panel.length[j]
                        )/b
                    )

                    g = np.arctan2(
                        e*self.Panel.length[j],
                        b + a*self.Panel.length[j]
                    )

                    p = (
                        (self.Panel.x_mid[i] - self.Panel.x[j])
                        * np.sin(
                            self.Panel.theta[i] - 2*self.Panel.theta[j]
                        )
                        +
                        (self.Panel.y_mid[i] - self.Panel.y[j])
                        * np.cos(
                            self.Panel.theta[i] - 2*self.Panel.theta[j]
                        )
                    )

                    q = (
                        (self.Panel.x_mid[i] - self.Panel.x[j])
                        * np.cos(
                            self.Panel.theta[i] - 2*self.Panel.theta[j]
                        )
                        -
                        (self.Panel.y_mid[i] - self.Panel.y[j])
                        * np.sin(
                            self.Panel.theta[i] - 2*self.Panel.theta[j]
                        )
                    )

                    cn2 = (
                        d
                        + 0.5*q*f/self.Panel.length[j]
                        - (a*c + d*e)*g/self.Panel.length[j]
                    )

                    cn1 = 0.5*d*f + c*g - cn2

                    ct2 = (
                        c
                        + 0.5*p*f/self.Panel.length[j]
                        + (a*d - c*e)*g/self.Panel.length[j]
                    )

                    ct1 = 0.5*c*f - d*g - ct2

                    Cn1[i, j] = cn1/(2*np.pi)
                    Cn2[i, j] = cn2/(2*np.pi)

                    Ct1[i, j] = ct1/(2*np.pi)
                    Ct2[i, j] = ct2/(2*np.pi)

        # Assemble system matrices

        An = np.zeros((N+1, N+1))
        At = np.zeros((N, N+1))

        RHS_full = np.zeros(N+1)

        An[:N, 0] = Cn1[:, 0]
        An[:N, N] = Cn2[:, N-1]

        At[:, 0] = Ct1[:, 0]
        At[:, N] = Ct2[:, N-1]

        for j in range(1, N):

            An[:N, j] = Cn1[:, j] + Cn2[:, j-1]

            At[:, j] = Ct1[:, j] + Ct2[:, j-1]

        RHS_full[:N] = RHS

        # Kutta condition

        An[N, 0] = 1
        An[N, N] = 1

        RHS_full[N] = 0

        self.An = An
        self.At = At

        # Solve nodal vortex strengths

        self.gamma1 = np.linalg.solve(An, RHS_full)

        # Tangential velocity

        self.Vt = (
            self.V_fs*np.cos(self.Panel.theta - self.alpha)
            + self.At @ self.gamma1
        )

        # Pressure coefficient

        self.Cp = 1 - (self.Vt/self.V_fs)**2

        # Lift coefficient

        c = self.Panel.x.max() - self.Panel.x.min()

        Gamma1 = np.sum(
            0.5*(self.gamma1[:-1] + self.gamma1[1:])
            * self.Panel.length
        )

        self.Cl = 2*Gamma1/(self.V_fs*c)

        return self.Cp, self.Cl
```

```python
# Generate circular cylinder

cyl = Cylinder(radius=1.0)

x_cyl, y_cyl = cyl.generate(num_points=100)

Panel_cyl = Panel(x_cyl, y_cyl)

# Apply constant-strength method

solver_cyl = Vortex_Panel_Method(
    Panel_cyl,
    V_fs=1.0,
    alpha=0.0
)

Cp_num, Cl_num = solver_cyl.constant_vortex_strength()

# Analytical solution

beta = np.arctan2(
    Panel_cyl.y_mid,
    Panel_cyl.x_mid
)

Cp_analytical = 1 - 4*np.sin(beta)**2

# Plot comparison

plt.figure(figsize=(9, 5))

plt.plot(
    np.degrees(beta),
    Cp_analytical,
    'k-',
    linewidth=2,
    label='Analytical Solution'
)

plt.plot(
    np.degrees(beta),
    Cp_num,
    'ro',
    markersize=4,
    label='Numerical Solution'
)

plt.xlabel("Angular Position (degrees)")
plt.ylabel("Pressure Coefficient")
plt.title("Pressure Distribution over Circular Cylinder")

plt.grid(True)
plt.legend()
plt.show()

print("Numerical Lift Coefficient =", Cl_num)
```

```python
# Airfoil generation

airfoil = Airfoil()

x_sym, y_sym = airfoil.symmetric(0.12, 100)

x_cam, y_cam = airfoil.cambered(0.12, 100)

# Flow conditions

V_freestream = 10
alpha_deg = 5

# Panel discretization

panel_sym = Panel(x_sym, y_sym)

panel_cam = Panel(x_cam, y_cam)

# NACA 0012 - Constant method

solver_sym_const = Vortex_Panel_Method(
    panel_sym, V_fs=V_freestream, alpha=alpha_deg
)

Cp_sym_const, Cl_sym_const = (
    solver_sym_const.constant_vortex_strength()
)

# NACA 0012 - Linear method

solver_sym_lin = Vortex_Panel_Method(
    panel_sym, V_fs=V_freestream, alpha=alpha_deg
)

Cp_sym_lin, Cl_sym_lin = (
    solver_sym_lin.linear_vortex_strength()
)

# NACA 2412 - Constant method

solver_cam_const = Vortex_Panel_Method(
    panel_cam, V_fs=V_freestream, alpha=alpha_deg
)

Cp_cam_const, Cl_cam_const = (
    solver_cam_const.constant_vortex_strength()
)

# NACA 2412 - Linear method

solver_cam_lin = Vortex_Panel_Method(
    panel_cam, V_fs=V_freestream, alpha=alpha_deg
)

Cp_cam_lin, Cl_cam_lin = (
    solver_cam_lin.linear_vortex_strength()
)

# Display lift coefficients

print("NACA 0012")
print("Constant strength Cl =", Cl_sym_const)
print("Linear strength Cl   =", Cl_sym_lin)

print("\nNACA 2412")
print("Constant strength Cl =", Cl_cam_const)
print("Linear strength Cl   =", Cl_cam_lin)
```

```python
N_values = [20, 40, 80, 160, 320]

Cl_results_const = []
Cl_results_lin = []

print("GRID CONVERGENCE RESULTS")

print(
    f"{'Points':<12} | "
    f"{'Constant Cl':<15} | "
    f"{'Linear Cl':<15}"
)

print("-" * 48)

for n in N_values:

    airfoil = Airfoil()

    x_cam, y_cam = airfoil.cambered(
        0.12,
        num_points=n
    )

    panel_cam = Panel(x_cam, y_cam)

    # Constant-strength method

    solver_const = Vortex_Panel_Method(
        panel_cam,
        V_fs=V_freestream,
        alpha=alpha_deg
    )

    _, Cl_const = solver_const.constant_vortex_strength()

    Cl_results_const.append(Cl_const)

    # Linear-strength method

    solver_lin = Vortex_Panel_Method(
        panel_cam,
        V_fs=V_freestream,
        alpha=alpha_deg
    )

    _, Cl_lin = solver_lin.linear_vortex_strength()

    Cl_results_lin.append(Cl_lin)

    print(
        f"{n:<12} | "
        f"{Cl_const:<15.5f} | "
        f"{Cl_lin:<15.5f}"
    )

# Plot convergence

plt.figure(figsize=(8, 5))

plt.plot(
    N_values,
    Cl_results_const,
    'bo-',
    label="Constant Strength"
)

plt.plot(
    N_values,
    Cl_results_lin,
    'r*-',
    label="Linear Strength"
)

plt.xlabel("Number of Coordinate Points")

plt.ylabel("Lift Coefficient")

plt.title("Grid Convergence Study - NACA 2412")

plt.grid(True)
plt.legend()

plt.show()
```
